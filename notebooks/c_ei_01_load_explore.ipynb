### Timing, two-tier validation, and resilient GeoParquet loading

Every stage below runs inside `step(...)` so slow calls surface immediately, and every data check records into `CHECKS` under one of two verdicts:

- **`check()` (`FAIL` when false)**: grades **our pipeline**, including 11-character `census_tract_geoid` keys, leading-zero FIPS preservation (`AL` = `'01'`, `CA` = `'06'`), graph vertex/edge primary-key uniqueness, and 100% foreign-key endpoint integrity.
- **`note()` (`WARN` up to `limit`, `FAIL` above it)**: flags **dataset traps** you need to know before building routing logic (such as the all-zero `limited_english_rate` column in this ACS extract).

When imported into Colab Enterprise by raw GitHub URL, `read_starter_parquet()` fetches the starter GeoParquet files directly over HTTPS (or via authenticated Cloud Storage fallback) and `load_table()` preserves spherical WKB metadata so `tract_geom`, `intake_geom`, and `provider_geom` land as native `GEOGRAPHY` columns.

````python
STEPS, CHECKS = {}, []
NOTEBOOK_START = time.time()

GITHUB_ORG = os.environ.get("A4I_GITHUB_ORG", "Bornism")
REPO_NAME = "A4I2026-challenge-early-intervention"
SNAPSHOT_BUCKET = os.environ.get("A4I_SNAPSHOT_BUCKET", "class-demo")
SNAPSHOT_PREFIX = "a4i-2026/challenge-early-intervention"
UA = {"User-Agent": "A4I2026-challenge-ei-notebook"}


@contextlib.contextmanager
def step(name):
    t0 = time.time()
    print(f"[{name}] ...")
    try:
        yield
    finally:
        dt = time.time() - t0
        STEPS[name] = round(dt, 1)
        print(f"[{name}] {dt:.1f}s")


def _record(name, verdict, detail):
    for c in CHECKS:
        if c["check"] == name:
            c["verdict"], c["detail"] = verdict, detail
            return
    CHECKS.append({"check": name, "verdict": verdict, "detail": detail})


def check(name, ok, detail=""):
    _record(name, "PASS" if ok else "FAIL", detail)
    print(f"  {'PASS' if ok else 'FAIL'}  {name}" + (f": {detail}" if detail else ""))
    return bool(ok)


def note(name, bad, total, detail="", limit=0.01, rows=None):
    frac = (bad / total) if total else 0.0
    verdict = "PASS" if bad == 0 else ("FAIL" if (total and frac > limit) else "WARN")
    msg = detail or (f"{bad:,} of {total:,}" if total else f"{bad:,}")
    _record(name, verdict, msg)
    print(f"  {verdict}  {name}: {msg}")
    if verdict != "PASS" and rows is not None:
        display(rows.head(10))
    return verdict


def safe_str(s, width=None):
    out = s.astype("string")
    if width:
        out = out.str.strip().str.zfill(width)
    return out.where(out.notna(), None)


def spread(series, label=""):
    s = pd.to_numeric(pd.Series(series), errors="coerce").dropna()
    if s.empty:
        return {}
    p10, p50, p90 = s.quantile([0.10, 0.50, 0.90])
    out = {
        "n": int(s.size),
        "min": round(float(s.min()), 3),
        "p10": round(float(p10), 3),
        "median": round(float(p50), 3),
        "p90": round(float(p90), 3),
        "max": round(float(s.max()), 3),
    }
    if label:
        print(f"  {label:<32} " + "  ".join(f"{k}={v}" for k, v in out.items()))
    return out
````

````python
def _fetch_parquet_bytes(url, tries=4, timeout=60):
    delay = 1.5
    for attempt in range(tries):
        try:
            resp = requests.get(url, timeout=timeout, headers=UA)
        except requests.RequestException as exc:
            if attempt == tries - 1:
                raise RuntimeError(f"{url} — {type(exc).__name__}: {exc}") from exc
            time.sleep(delay)
            delay *= 2
            continue
        if resp.status_code == 429 or 500 <= resp.status_code < 600:
            if attempt == tries - 1:
                raise RuntimeError(f"{url} — HTTP {resp.status_code} after {tries} tries")
            time.sleep(delay)
            delay *= 2
            continue
        if resp.status_code != 200:
            raise RuntimeError(f"{url} — HTTP {resp.status_code}")
        if resp.content[:4] != b"PAR1":
            raise RuntimeError(f"{url} — response is not a valid Parquet file")
        return resp.content
    raise RuntimeError(f"{url} — exhausted retries")


def _fetch_gcs_parquet_bytes(bucket_name, blob_path):
    from google.cloud import storage as gcs_storage
    client = gcs_storage.Client(project=PROJECT)
    blob = client.bucket(bucket_name).blob(blob_path)
    data = blob.download_as_bytes(timeout=60)
    if data[:4] != b"PAR1":
        raise RuntimeError(f"gs://{bucket_name}/{blob_path} is not a valid Parquet file")
    return data


def read_starter_parquet(filename):
    if not filename or "/" in filename or ".." in filename:
        raise ValueError(f"Invalid starter filename: {filename!r}")
    data_dir = os.environ.get("A4I_DATA_DIR", "").strip()
    local_candidates = (
        ([Path(data_dir) / filename] if data_dir else [])
        + [Path("../data") / filename, Path("data") / filename, Path(filename)]
    )
    for path in local_candidates:
        if path.is_file():
            return pd.read_parquet(path)

    org_candidates = list(dict.fromkeys([GITHUB_ORG, "Bornism", "haggman"]))
    base_override = os.environ.get("A4I_DATA_BASE_URL", "").strip().rstrip("/")
    urls = ([f"{base_override}/{filename}"] if base_override else []) + [
        f"https://raw.githubusercontent.com/{org}/{REPO_NAME}/main/data/{filename}"
        for org in org_candidates
    ]
    errors = []
    for url in urls:
        try:
            return pd.read_parquet(io.BytesIO(_fetch_parquet_bytes(url)))
        except RuntimeError as exc:
            errors.append(str(exc))

    if SNAPSHOT_BUCKET:
        blob_path = f"{SNAPSHOT_PREFIX}/starter/{filename}"
        try:
            return pd.read_parquet(io.BytesIO(_fetch_gcs_parquet_bytes(SNAPSHOT_BUCKET, blob_path)))
        except Exception as exc:  # noqa: BLE001
            errors.append(f"gs://{SNAPSHOT_BUCKET}/{blob_path} ({type(exc).__name__}: {exc})")

    raise RuntimeError(f"Could not locate {filename} locally or remotely: {'; '.join(errors)}")

def _to_geoparquet_buffer(df, geom_col=None):
    table = pa.Table.from_pandas(df, preserve_index=False)
    if geom_col and geom_col in df.columns:
        geo_meta = {
            "version": "1.0.0",
            "primary_column": geom_col,
            "columns": {
                geom_col: {
                    "encoding": "WKB",
                    "geometry_types": [],
                    "crs": None,
                    "orientation": "counterclockwise",
                    "edges": "spherical",
                }
            },
        }
        meta = dict(table.schema.metadata or {})
        meta[b"geo"] = json.dumps(geo_meta).encode("utf-8")
        table = table.replace_schema_metadata(meta)
    buf = io.BytesIO()
    pq.write_table(table, buf)
    buf.seek(0)
    return buf


def load_table(df, table_name, geom_col=None, description=""):
    if df is None or not table_name:
        raise ValueError("df and table_name are required")
    ref = f"{PROJECT}.{DATASET}.{table_name}"
    buf = _to_geoparquet_buffer(df, geom_col=geom_col)
    cfg = bigquery.LoadJobConfig(
        source_format=bigquery.SourceFormat.PARQUET,
        write_disposition=bigquery.WriteDisposition.WRITE_TRUNCATE,
    )
    bq.load_table_from_file(buf, ref, job_config=cfg).result()
    tbl = bq.get_table(ref)
    if description:
        tbl.description = description
        bq.update_table(tbl, ["description"])
    print(f"  loaded {table_name:<38} {tbl.num_rows:>6,} rows")
    return tbl
````

### 1b. Provision a zero-baseline Enterprise reservation for BigQuery Graph

**Do not skip this cell.** BigQuery Property Graph queries (`GRAPH ... MATCH ...` and `GRAPH_TABLE`) require a reservation on the **Enterprise** or **Enterprise Plus** edition. Fresh Google Cloud projects default to on-demand pricing, which fails on the first `GRAPH` query with:

> *BigQuery Graph queries require a reservation with Enterprise or Enterprise Plus edition.*

The cell below creates an idempotent reservation (`a4i-graph`) in `region-us` with **`slot_capacity = 0` and `autoscale_max_slots = 50`**—meaning it costs **\$0 while idle** and borrows a handful of slots for a few seconds only when a graph query runs. Re-running is a fast no-op once in place.

````python
RESERVATION = "a4i-graph"
_region = f"region-{LOCATION.lower()}"
_res_path = f"`{PROJECT}.{_region}.{RESERVATION}`"
_asg_path = f"`{PROJECT}.{_region}.{RESERVATION}.{RESERVATION}-query`"


def _enable_reservation_api():
    import subprocess
    import google.auth
    from google.auth.transport.requests import AuthorizedSession

    print(f"  enabling bigqueryreservation.googleapis.com on {PROJECT}...")
    enabled = False
    try:
        creds, _ = google.auth.default(scopes=["https://www.googleapis.com/auth/cloud-platform"])
        authed = AuthorizedSession(creds)
        url = f"https://serviceusage.googleapis.com/v1/projects/{PROJECT}/services/bigqueryreservation.googleapis.com:enable"
        resp = authed.post(url, json={}, timeout=60)
        if resp.status_code in (200, 201):
            op = resp.json()
            op_name = op.get("name", "")
            for _ in range(24):
                if op.get("done", True) or op_name.endswith("DONE_OPERATION"):
                    break
                time.sleep(5)
                poll = authed.get(f"https://serviceusage.googleapis.com/v1/{op_name}", timeout=30)
                if poll.status_code == 200:
                    op = poll.json()
            enabled = True
    except Exception:  # noqa: BLE001
        pass

    if not enabled:
        subprocess.run(
            ["gcloud", "services", "enable", "bigqueryreservation.googleapis.com", f"--project={PROJECT}", "--quiet"],
            check=False,
            capture_output=True,
        )
    time.sleep(10)


def _reservation_ddl(sql, what):
    for attempt in range(2):
        try:
            bq.query(sql).result()
            print(f"  created {what}")
            return True
        except Exception as exc:  # noqa: BLE001
            msg = str(exc)
            if "already" in msg.lower():
                print(f"  {what}: already in place")
                return True  # Treat as successful since it already exists
            if "quota" in msg.lower() or "limit" in msg.lower() or "exceeded" in msg.lower():
                print(f"  WARNING: Quota limit hit while trying to create {what}. Skipping.")
                return False
            if attempt == 0 and ("bigqueryreservation" in msg or "has not been used" in msg or "is disabled" in msg):
                _enable_reservation_api()
                continue
            if "bigqueryreservation" in msg or "has not been used" in msg or "is disabled" in msg:
                raise RuntimeError(
                    "BigQuery Reservation API is still propagating. Wait 30 seconds and re-run this cell, or run in Cloud Shell:\n"
                    f"  gcloud services enable bigqueryreservation.googleapis.com --project={PROJECT}"
                ) from exc
            raise
    return False


with step("1b graph-capable Enterprise reservation"):
    if os.environ.get("A4I_SKIP_RESERVATION") == "1":
        print("  skipped reservation DDL (A4I_SKIP_RESERVATION=1)")
    else:
        # Only attempt query assignment if reservation creation succeeded or already exists
        created = _reservation_ddl(
            f"CREATE RESERVATION {_res_path} "
            f"OPTIONS (edition = 'ENTERPRISE', slot_capacity = 0, autoscale_max_slots = 50)",
            f"reservation {RESERVATION} (ENTERPRISE, 0 baseline slots, autoscale max 50)",
        )
        if created:
            assigned = _reservation_ddl(
                f"CREATE ASSIGNMENT {_asg_path} "
                f"OPTIONS (assignee = 'projects/{PROJECT}', job_type = 'QUERY')",
                f"QUERY assignment for {PROJECT}",
            )
            if assigned:
                print("  waiting 60s for new reservation assignment to propagate...")
                time.sleep(60)
            print(f"  graph queries in {PROJECT} assigned to {RESERVATION}")
        else:
            print("  Skipping query assignment because the Enterprise reservation could not be created.")
````

## 2. Start with the wrong answer

Suppose a state Part C director asks: **"Where is our referral network breaking the federal 45-day IFSP timeline, and where should we intervene first?"**

The obvious flat-SQL instinct is to sort `pediatric_and_ei_provider_directory` by `active_waitlist_count DESC` and send emergency funding to whichever hub sits at the top. Let's run that flat sort right alongside a multi-hop episode trace on `{STATE}` and watch what flat SQL misses:

- Sorting by `active_waitlist_count` points at saturated downstream `Multidisciplinary_Eval_Hub` queues (waitlist `60–81`), which are **symptoms** at hop 2 and hop 3.
- Tracing complete `case_chain_id` trajectories reveals that `Part_C_Intake_Coordinator` nodes look completely healthy in a flat table (`active_waitlist_count = 20–29`, `median_internal_wait_days = 5–6`), yet **route families via `paper_fax` into already-saturated or non-bilingual evaluation hubs**—triggering 3- and 4-hop re-referral bounces (`70–75` total wait days, blowing past the **45-day federal statutory ceiling**)!

````python
with step("section 2 flat-waitlist vs multi-hop episode hook"):
    _prov = read_starter_parquet("pediatric_and_ei_provider_directory.parquet")
    _logs = read_starter_parquet("deidentified_part_c_referral_logs.parquet")
    _sp = _prov[_prov["state_code"] == STATE].copy()
    _sl = _logs[_logs["state_code"] == STATE].copy()

    print(f"1) Naive flat sort — top 5 providers in {STATE} by active_waitlist_count:")
    display(
        _sp.sort_values("active_waitlist_count", ascending=False)[
            [
                "provider_id",
                "node_role",
                "active_waitlist_count",
                "median_internal_wait_days",
                "bilingual_services_listed",
            ]
        ].head(5)
    )

    _episodes = (
        _sl.groupby("case_chain_id")
        .agg(total_hops=("hop_sequence", "count"), episode_wait_days=("handoff_wait_days", "sum"))
        .reset_index()
    )
    _hop1 = _sl[_sl["hop_sequence"] == 1].merge(_episodes, on="case_chain_id")
    _coord_impact = (
        _hop1.groupby("target_provider_id")
        .agg(
            routed_episodes=("case_chain_id", "count"),
            chains_3plus_hops=("total_hops", lambda s: int((s >= 3).sum())),
            chains_over_45d=("episode_wait_days", lambda s: int((s > 45).sum())),
            mean_episode_wait_d=("episode_wait_days", lambda s: round(float(s.mean()), 1)),
        )
        .reset_index()
        .merge(
            _sp[["provider_id", "node_role", "active_waitlist_count", "median_internal_wait_days"]],
            left_on="target_provider_id",
            right_on="provider_id",
        )
        .sort_values(["chains_over_45d", "mean_episode_wait_d"], ascending=False)
    )
    print(f"\n2) Multi-hop reality — Part_C_Intake_Coordinator hubs ranked by >45-day breaches:")
    display(
        _coord_impact[
            [
                "provider_id",
                "active_waitlist_count",
                "median_internal_wait_days",
                "routed_episodes",
                "chains_3plus_hops",
                "chains_over_45d",
                "mean_episode_wait_d",
            ]
        ]
    )
````

Look at the bottom Coordinator row (`SYN-PROV-*-012` / `011`): an internal wait of **5–6 days** and a modest waitlist, yet **9 to 10 of its 11 routed episodes** bounce across 3–4 hops and breach the 45-day IDEA Part C deadline! That multi-hop path dependency is why **BigQuery Property Graphs (`CREATE PROPERTY GRAPH` + ISO `GQL`)** is your differentiator.

## 4. Census under-5 tract demographics (`acs_under5_tract_demographics`) — and an all-zero column trap!

This table comes from the **US Census Bureau ACS 5-Year** estimates (`children_under_5_count`, `poverty_rate`, `no_vehicle_rate`, `assistance_rate`, `tract_geom`, and post-output `audit_*` shares).

**Inspect your distributions before writing formulas.** Run `spread()` below and look at `limited_english_rate` (`limited_english_speaking_pop_count`): across all tracts in this ACS extract, **every value is `0.0`**! If your team writes a composite neighborhood triage score like `0.4 * poverty_rate + 0.4 * limited_english_rate + 0.2 * no_vehicle_rate` without checking `spread()`, `40%` of your formula silently collapses to zero—while real language screening disparities sit in `nsch_developmental_screening_survey` (`primary_home_language`) and `pediatric_and_ei_provider_directory` (`bilingual_services_listed`)! ## 10. Episode breach & spatial bilingual-access payoff queries

Before writing GQL in Section 13, let's quantify two foundational truths in SQL:

1. **How hop depth drives federal 45-day IDEA Part C breaches:** 2-hop chains never breach 45 days; 3-hop chains sit right on the razor's edge; 4-hop re-referral bounce chains breach **100% of the time**.
2. **Where high-poverty under-5 neighborhoods sit farthest from a bilingual, Medicaid-accepting `Multidisciplinary_Eval_Hub`:** combining `ST_CENTROID(tract_geom)` with `provider_geom`.

````python
with step("episode hop-depth & bilingual distance payoff queries"):
    hop_breaches = bq.query(f"""
        WITH episodes AS (
            SELECT
                case_chain_id,
                COUNT(*) AS hops,
                SUM(handoff_wait_days) AS total_wait_days,
                COUNTIF(transmission_channel = 'paper_fax') AS fax_hops,
                MAX(IF(STARTS_WITH(handoff_reason, 'Re-Referral Bounce'), 1, 0)) AS had_bounce
            FROM `{PROJECT}.{DATASET}.deidentified_part_c_referral_logs`
            GROUP BY case_chain_id
        )
        SELECT
            hops,
            COUNT(*) AS episodes,
            ROUND(AVG(total_wait_days), 1) AS avg_episode_wait_days,
            MIN(total_wait_days) AS min_wait_d,
            MAX(total_wait_days) AS max_wait_d,
            COUNTIF(total_wait_days > 45) AS breaches_over_45d,
            SUM(had_bounce) AS bounce_episodes,
            ROUND(AVG(fax_hops), 1) AS avg_fax_hops
        FROM episodes
        GROUP BY hops
        ORDER BY hops
    """).to_dataframe()
    print("IDEA Part C 45-day compliance by referral chain length:")
    display(hop_breaches)

    eval_gaps = bq.query(f"""
        SELECT
            d.census_tract_geoid,
            d.county_name,
            d.children_under_5_count,
            ROUND(d.poverty_rate * 100, 1) AS poverty_pct,
            ROUND(d.no_vehicle_rate * 100, 1) AS no_vehicle_pct,
            ROUND(
                MIN(ST_DISTANCE(ST_CENTROID(d.tract_geom), p.provider_geom)) / 1000.0,
                1
            ) AS km_to_bilingual_medicaid_eval_hub
        FROM `{PROJECT}.{DATASET}.acs_under5_tract_demographics` AS d
        CROSS JOIN `{PROJECT}.{DATASET}.pediatric_and_ei_provider_directory` AS p
        WHERE p.node_role = 'Multidisciplinary_Eval_Hub'
          AND p.accepts_medicaid = TRUE
          AND p.bilingual_services_listed = TRUE
        GROUP BY 1, 2, 3, 4, 5
        ORDER BY km_to_bilingual_medicaid_eval_hub DESC, d.children_under_5_count DESC
        LIMIT 8
    """).to_dataframe()
    print("\nUnder-5 tracts farthest from a bilingual, Medicaid-accepting Evaluation Hub:")
    display(eval_gaps)
````

### 10b. Zero-Vehicle Transit Accessibility Guardrail (`tract_transit_guardrails`)

For caregivers in tracts with high zero-vehicle rates (`no_vehicle_rate >= 15%`), evaluate physical distance to the nearest bilingual Medicaid evaluation hub. If distance exceeds 5 km, the system flags a transit desert and triggers an automated recommendation for a **Mobile Developmental Evaluation Team** or **Telehealth Intake**.

> **State Scope Note:**  
> The `tract_transit_guardrails` table below is compiled for **Mississippi (MS)**, our challenge reference state. Because the query relies entirely on standardized BigQuery `GEOGRAPHY` and ACS schema definitions, executing this cell against any of the other five states (`AL`, `GA`, `TX`, `CA`, `NY`) will dynamically compile the transit risk tiers for that state with zero code modifications.

````python
with step("create tract_transit_guardrails table"):
    bq.query(f"""
        CREATE OR REPLACE TABLE `{PROJECT}.{DATASET}.tract_transit_guardrails` AS
        SELECT
            d.census_tract_geoid,
            d.county_name,
            d.state_code,
            ROUND(d.poverty_rate * 100, 1) AS poverty_pct,
            ROUND(d.no_vehicle_rate * 100, 1) AS no_vehicle_pct,
            p.provider_id AS nearest_eval_hub_id,
            p.synthetic_organization_name AS nearest_bilingual_eval_hub,
            ROUND(ST_DISTANCE(ST_CENTROID(d.tract_geom), p.provider_geom) / 1000.0, 1) AS distance_km,
            CASE
                WHEN d.no_vehicle_rate >= 0.15 AND ST_DISTANCE(ST_CENTROID(d.tract_geom), p.provider_geom) > 5000
                    THEN 'HIGH_TRANSIT_BARRIER'
                WHEN d.no_vehicle_rate >= 0.15
                    THEN 'LOCAL_OR_WALKABLE'
                ELSE 'PERSONAL_VEHICLE_ASSUMED'
            END AS transit_risk_tier,
            IF(d.no_vehicle_rate >= 0.15 AND ST_DISTANCE(ST_CENTROID(d.tract_geom), p.provider_geom) > 5000,
               'Dispatch mobile evaluation van or schedule tele-intake',
               'Standard on-site clinic appointment') AS coordinator_action
        FROM `{PROJECT}.{DATASET}.acs_under5_tract_demographics` d
        CROSS JOIN `{PROJECT}.{DATASET}.pediatric_and_ei_provider_directory` p
        WHERE p.node_role = 'Multidisciplinary_Eval_Hub'
          AND p.accepts_medicaid = TRUE
          AND p.bilingual_services_listed = TRUE
        QUALIFY ROW_NUMBER() OVER (
            PARTITION BY d.census_tract_geoid
            ORDER BY ST_DISTANCE(ST_CENTROID(d.tract_geom), p.provider_geom) ASC
        ) = 1;
    """).result()

    preview = bq.query(f"""
        SELECT
            census_tract_geoid,
            county_name,
            no_vehicle_pct,
            distance_km,
            transit_risk_tier,
            coordinator_action
        FROM `{PROJECT}.{DATASET}.tract_transit_guardrails`
        ORDER BY no_vehicle_pct DESC, distance_km DESC
        LIMIT 6;
    """).to_dataframe()

    print("Top high-transit-risk tracts in Mississippi:")
    display(preview)
````

### 10c. Counterfactual simulation: intake coordinator transmission upgrades

What happens to federal compliance if regional intake hubs stop relying on paper fax? This counterfactual simulation models a targeted operational fix: converting `paper_fax` referral transmissions (which average ~19.5 days per handoff) to electronic portal intake (`ehr_electronic_portal`, calibrated to a 4-day baseline). We recalculate cumulative episode durations across all historical chains to quantify both the reduction in 45-day breaches and the total child-days saved.

````python
with step("counterfactual bottleneck simulation"):
    sim_results = bq.query(f"""
        WITH simulation AS (
            SELECT
                case_chain_id,
                -- Actual baseline wait days for the referral episode
                SUM(handoff_wait_days) AS baseline_wait_days,
                -- Simulated wait days: replace paper_fax with electronic intake (4 days)
                SUM(IF(transmission_channel = 'paper_fax', 4, handoff_wait_days)) AS simulated_wait_days
            FROM `{PROJECT}.{DATASET}.deidentified_part_c_referral_logs`
            GROUP BY case_chain_id
        )
        SELECT
            COUNT(*) AS total_episodes,
            COUNTIF(baseline_wait_days > 45) AS baseline_breaches,
            COUNTIF(simulated_wait_days <= 45 AND baseline_wait_days > 45) AS cases_recovered_to_compliance,
            ROUND(COUNTIF(simulated_wait_days <= 45 AND baseline_wait_days > 45) * 100.0 / COUNTIF(baseline_wait_days > 45), 1) AS recovery_rate_pct,
            SUM(baseline_wait_days - simulated_wait_days) AS total_child_days_saved
        FROM simulation;
    """).to_dataframe()
    
    print("Counterfactual Network Fix — Simulated Results:")
    display(sim_results)
````

### 10d. Part C age-out risk tiers (36-month transition cliff)

Under federal statute, Part C early intervention services strictly end on a child's third birthday (36 months), at which point jurisdiction transfers to school district Part B preschool special education. When a toddler enters referral at 30 to 34 months, standard 45-day evaluation windows leave almost no room for therapy delivery before program exit. We stratify incoming referrals by age at entry to flag cases requiring expedited multidisciplinary evaluations and proactive 60-day Part B transition conferences.

````python
with step("age-out cliff transition triage"):
    age_out_triage = bq.query("""
        SELECT
            child_age_months,
            CASE 
                WHEN child_age_months >= 32 THEN 'CRITICAL_AGE_OUT_CLIFF'
                WHEN child_age_months >= 30 THEN 'ELEVATED_TRANSITION_RISK'
                ELSE 'STANDARD_UNDER_3'
            END AS transition_risk_tier,
            CASE 
                WHEN child_age_months >= 32 THEN 'Fast-track expedited evaluation: parallel Part B preschool notification required'
                WHEN child_age_months >= 30 THEN 'Flag 60-day Part B transition conference upon initial IFSP scheduling'
                ELSE 'Standard Part C 45-day evaluation workflow'
            END AS clinical_action_guidance
        FROM UNNEST([18, 22, 28, 30, 32, 34]) AS child_age_months
        ORDER BY child_age_months DESC;
    """).to_dataframe()
    
    print("Part C 36-Month Transition Risk Tiers:")
    display(age_out_triage)
````

## 11. Deliberate exclusions & the Post-Output Equity Audit rule

1. **`audit_*` columns (`audit_black_share`, `audit_hispanic_share`, `audit_asian_share`, `audit_indigenous_share`, `audit_white_share`) are strictly banned from routing rules, priority scores, and LLM triage prompts.** Route children and prioritize network investments strictly on clinical referral urgency (`child_age_months` approaching the 36-month Part C age-out cliff!), wait-time bottlenecks, bilingual/Medicaid slot capacity, poverty, and travel distance.
2. **Join `audit_*` *after* generating your routing & capacity recommendations** to verify whether children living in high-poverty, zero-vehicle, and historically underserved communities experience disproportionate `>45-day` referral chains—and whether your proposed fix closes that gap.

````python
with step("post-output equity audit on >45d breached home tracts"):
    breach_equity = bq.query(f"""
        WITH chain_totals AS (
            SELECT
                case_chain_id,
                ANY_VALUE(child_home_tract_geoid) AS child_home_tract_geoid,
                SUM(handoff_wait_days) AS episode_wait_days
            FROM `{PROJECT}.{DATASET}.deidentified_part_c_referral_logs`
            GROUP BY case_chain_id
        )
        SELECT
            IF(c.episode_wait_days > 45, 'Breached (>45 days)', 'Timely (<=45 days)') AS ifsp_status,
            COUNT(*) AS episodes,
            ROUND(AVG(d.poverty_rate) * 100, 1) AS avg_tract_poverty_pct,
            ROUND(AVG(d.no_vehicle_rate) * 100, 1) AS avg_no_vehicle_pct,
            ROUND(AVG(d.audit_black_share) * 100, 1) AS audit_black_pct,
            ROUND(AVG(d.audit_hispanic_share) * 100, 1) AS audit_hispanic_pct,
            ROUND(AVG(d.audit_white_share) * 100, 1) AS audit_white_pct
        FROM chain_totals AS c
        INNER JOIN `{PROJECT}.{DATASET}.acs_under5_tract_demographics` AS d
            ON c.child_home_tract_geoid = d.census_tract_geoid
        GROUP BY 1
        ORDER BY 1
    """).to_dataframe()
    display(breach_equity)
````

## 13. Framing your differentiator — BigQuery Property Graphs (`CREATE PROPERTY GRAPH` + ISO `GQL`)

This is where the starter notebook stops and your team's graph engineering and agent design take over.

### Step A — Declare the homogeneous `CareNode` Property Graph

Notice **two critical syntax rules** in the DDL below:
1. Table references inside `NODE TABLES` and `EDGE TABLES` are **dataset-qualified and unbackticked** (`a4i_early_intervention.pediatric_and_ei_provider_directory`), while `REFERENCES CareNode (provider_id)` uses the **alias without the dataset**.
2. Exclude `provider_geom` (`PROPERTIES ALL COLUMNS EXCEPT (provider_geom)`) so every graph property is a scalar GQL-compatible type (`latitude` and `longitude` remain on `CareNode`!).

```sql
CREATE OR REPLACE PROPERTY GRAPH a4i_early_intervention.ReferralGraph
  NODE TABLES (
    a4i_early_intervention.pediatric_and_ei_provider_directory AS CareNode
      KEY (provider_id)
      LABEL CareNode
      PROPERTIES ALL COLUMNS EXCEPT (provider_geom)
  )
  EDGE TABLES (
    a4i_early_intervention.deidentified_part_c_referral_logs AS ReferralHandoff
      KEY (referral_event_id)
      SOURCE KEY (source_provider_id) REFERENCES CareNode (provider_id)
      DESTINATION KEY (target_provider_id) REFERENCES CareNode (provider_id)
      LABEL ReferralHandoff
      PROPERTIES ALL COLUMNS
  );
```

### Step B — Variable-length `-{2,4}->` ISO GQL traversal (not Cypher!)

BigQuery implements **ISO/IEC 39075 GQL**—not Neo4j Cypher, and not Gremlin. Look closely at line 6 below (`ALL_DIFFERENT` / `case_chain_id` correlation): without correlating `case_chain_id` across every hop in `hops`, a `-{2,4}->` pattern connects Child A's hop-1 referral into a coordinator with Child B's hop-2 referral out of that coordinator!

```sql
GRAPH a4i_early_intervention.ReferralGraph
MATCH (clinic:CareNode {node_role: 'Screening_Pediatric_Clinic'})
      -[hops:ReferralHandoff]->{2,4}
      (vendor:CareNode {node_role: 'EI_Therapy_Vendor'})
-- Loaded tables already hold your selected STATE (add AND clinic.state_code = 'CA' etc. if multi-state)
WHERE ARRAY_LENGTH(hops) >= 3
  AND (SELECT COUNT(DISTINCT h.case_chain_id) FROM UNNEST(hops) AS h) = 1
LET total_hops = ARRAY_LENGTH(hops),
    chain_id = hops[OFFSET(0)].case_chain_id,
    chain_wait_days = (SELECT SUM(h.handoff_wait_days) FROM UNNEST(hops) AS h),
    fax_hops = (SELECT COUNTIF(h.transmission_channel = 'paper_fax') FROM UNNEST(hops) AS h)
FILTER chain_wait_days > 45
RETURN
  chain_id,
  clinic.synthetic_organization_name AS origin_clinic,
  hops[OFFSET(0)].target_provider_id AS intake_coordinator_id,
  vendor.synthetic_organization_name AS final_vendor,
  total_hops,
  fax_hops,
  chain_wait_days
ORDER BY chain_wait_days DESC
LIMIT 15;
```

### Seven BigQuery Graph traps — save yourself 45 minutes

1. **Correlate `case_chain_id` across `UNNEST(hops)` when tracing real child episodes.** Use `(SELECT COUNT(DISTINCT h.case_chain_id) FROM UNNEST(hops) AS h) = 1` for historical episode audits—and drop that constraint deliberately *only* when asking "what alternative fast-track paths *could* this coordinator route a new child onto?"
2. **There is no `HAVING` in ISO GQL.** Compute per-path values with `LET` or aggregate in `RETURN`, then chain with `NEXT` + `FILTER`:
   ```sql
   GRAPH a4i_early_intervention.ReferralGraph
   MATCH (c:CareNode {node_role: 'Part_C_Intake_Coordinator'})-[e:ReferralHandoff]->(hub:CareNode)
   RETURN c.provider_id AS coord_id, hub.provider_id AS hub_id, AVG(e.handoff_wait_days) AS avg_wait
   NEXT
   FILTER avg_wait > 18
   RETURN coord_id, hub_id, ROUND(avg_wait, 1) AS avg_wait
   ORDER BY avg_wait DESC;
   ```
3. **Unbackticked `dataset.table` in `CREATE PROPERTY GRAPH`.** Backticking `` `project.dataset.table` `` turns the whole three-part name into one quoted identifier and makes `REFERENCES` fail with *"The referenced node table is not defined"*.
4. **Calling GQL from Python / ADK tools:** Wrap GQL inside standard SQL via `FROM GRAPH_TABLE(a4i_early_intervention.ReferralGraph MATCH ... COLUMNS (...))` or run `GRAPH a4i_early_intervention.ReferralGraph ...` directly via `bq.query(sql)` (works seamlessly once `Section 1b`'s `a4i-graph` reservation is active!).
5. **Visualizing paths requires `ANY SHORTEST` + `TO_JSON(p)`.** Returning bare columns renders a table; returning `TO_JSON(p) AS paths` over `MATCH p = ANY SHORTEST ...` keeps payloads well under Colab's 2 MB graph widget cap.
6. **Watch the 36-month Part C age-out cliff (`child_age_months`).** A 31-month-old toddler stuck in a 75-day fax bounce chain does not merely get a late IFSP—they risk aging out of Part C (`36 months`) before therapy starts!
7. **Spanner Graph docs look almost identical to BigQuery Graph docs.** Verify documentation URLs contain `/bigquery/`, not `/spanner/`.

### Four questions judges will ask from your 3-minute video & deck

1. **Graph necessity:** Show a bottleneck or fast-track reroute your GQL traversal uncovered that flat single-table sorting (`Section 2`) missed.
2. **Actionable routing vs. retrospective blame:** For a newly referred 28-month-old Spanish-speaking Medicaid toddler in a high-poverty tract, which specific `Clinic -> Coordinator -> Eval Hub -> Therapy Vendor` path keeps expected total wait under **45 days**, and why?
3. **Synthetic & grain honesty:** Does your deck/video clearly state that `SYN-PROV-*` / `SYN-PARTC-*` are `100% [Synthetic]` (protecting real clinicians) and that HRSA NSCH (`nsch_developmental_screening_survey`) is `Statewide`-only?
4. **Post-output Equity Audit:** What did joining `audit_*` and `no_vehicle_rate` reveal about which neighborhoods suffer `>45-day` breaches—and how did you address zero-vehicle travel barriers without using race as a routing input?

````python
# Section 13 - Step B: Run the 2..4 hop traversal via GQL (with SQL fallback if Enterprise reservation is unavailable)
gql_query = f"""
GRAPH {DATASET}.ReferralGraph
MATCH (clinic:CareNode {{node_role: 'Screening_Pediatric_Clinic'}})
      -[hops:ReferralHandoff]->{{2,4}}
      (vendor:CareNode {{node_role: 'EI_Therapy_Vendor'}})
WHERE ARRAY_LENGTH(hops) >= 3
  AND (SELECT COUNT(DISTINCT h.case_chain_id) FROM UNNEST(hops) AS h) = 1
LET total_hops = ARRAY_LENGTH(hops),
    chain_id = hops[OFFSET(0)].case_chain_id,
    chain_wait_days = (SELECT SUM(h.handoff_wait_days) FROM UNNEST(hops) AS h),
    fax_hops = (SELECT COUNTIF(h.transmission_channel = 'paper_fax') FROM UNNEST(hops) AS h)
FILTER chain_wait_days > 45
RETURN
  chain_id,
  clinic.synthetic_organization_name AS origin_clinic,
  hops[OFFSET(0)].target_provider_id AS intake_coordinator_id,
  vendor.synthetic_organization_name AS final_vendor,
  total_hops,
  fax_hops,
  chain_wait_days
ORDER BY chain_wait_days DESC
LIMIT 15;
"""

# Equivalent Standard SQL query in case Section 1b's Enterprise reservation quota limit blocks GQL
fallback_sql_query = f"""
WITH chain_hops AS (
  SELECT
    l.case_chain_id AS chain_id,
    ARRAY_AGG(l ORDER BY l.hop_sequence ASC) AS hops,
    COUNT(*) AS total_hops,
    COUNTIF(l.transmission_channel = 'paper_fax') AS fax_hops,
    SUM(l.handoff_wait_days) AS chain_wait_days
  FROM `{PROJECT}.{DATASET}.deidentified_part_c_referral_logs` AS l
  GROUP BY l.case_chain_id
  HAVING total_hops BETWEEN 3 AND 4
     AND chain_wait_days > 45
)
SELECT
  c.chain_id,
  clinic.synthetic_organization_name AS origin_clinic,
  c.hops[OFFSET(0)].target_provider_id AS intake_coordinator_id,
  vendor.synthetic_organization_name AS final_vendor,
  c.total_hops,
  c.fax_hops,
  c.chain_wait_days
FROM chain_hops AS c
JOIN `{PROJECT}.{DATASET}.pediatric_and_ei_provider_directory` AS clinic
  ON c.hops[OFFSET(0)].source_provider_id = clinic.provider_id
JOIN `{PROJECT}.{DATASET}.pediatric_and_ei_provider_directory` AS vendor
  ON c.hops[ARRAY_LENGTH(c.hops) - 1].target_provider_id = vendor.provider_id
WHERE clinic.node_role = 'Screening_Pediatric_Clinic'
  AND vendor.node_role = 'EI_Therapy_Vendor'
ORDER BY c.chain_wait_days DESC
LIMIT 15;
"""

try:
    print("Attempting ISO GQL query...")
    df_breaches = bq.query(gql_query).to_dataframe()
    print("SUCCESS: Executed via ISO GQL!")
except Exception as e:
    print(f"GQL query blocked (likely due to Section 1b reservation quota: {e}).")
    print("\nFalling back to equivalent Standard SQL multi-hop array query...")
    df_breaches = bq.query(fallback_sql_query).to_dataframe()

display(df_breaches)
````

````python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.lines import Line2D
import numpy as np

# 1. Fetch nodes, aggregated edges, and 36-month Age-Out Cliff stats
nodes_df = bq.query(f"""
    SELECT provider_id, node_role, active_waitlist_count, bilingual_services_listed, accepts_medicaid
    FROM `{PROJECT}.{DATASET}.pediatric_and_ei_provider_directory`
    ORDER BY node_role, provider_id
""").to_dataframe()

edges_df = bq.query(f"""
    SELECT
        source_id,
        source_role,
        target_id,
        target_role,
        COUNT(*) AS handoff_count,
        ROUND(AVG(wait_days), 1) AS avg_wait,
        COUNTIF(channel = 'paper_fax') AS fax_count,
        LOGICAL_OR(channel = 'paper_fax') AS has_fax,
        LOGICAL_OR(STARTS_WITH(reason, 'Re-Referral Bounce')) AS is_bounce
    FROM GRAPH_TABLE(
        {DATASET}.ReferralGraph
        MATCH (src:CareNode)-[e:ReferralHandoff]->(dst:CareNode)
        COLUMNS (
            src.provider_id AS source_id,
            src.node_role AS source_role,
            dst.provider_id AS target_id,
            dst.node_role AS target_role,
            e.handoff_wait_days AS wait_days,
            e.transmission_channel AS channel,
            e.handoff_reason AS reason
        )
    )
    GROUP BY source_id, source_role, target_id, target_role
""").to_dataframe()

# Check how many breached chains push toddlers past the 36-month IDEA Part C age-out cliff!
cliff_df = bq.query(f"""
    WITH ep AS (
        SELECT
            case_chain_id,
            MAX(child_age_months) AS age_m,
            SUM(handoff_wait_days) AS wait_d,
            ARRAY_AGG(target_provider_id ORDER BY hop_sequence DESC LIMIT 1)[OFFSET(0)] AS final_vendor_id
        FROM `{PROJECT}.{DATASET}.deidentified_part_c_referral_logs`
        GROUP BY case_chain_id
    )
    SELECT
        COUNTIF(wait_d > 45) AS breached_45d,
        COUNTIF(wait_d > 45 AND (age_m + wait_d / 30.4375) >= 36.0) AS age_out_cliff_cases,
        ARRAY_AGG(DISTINCT IF(wait_d > 45 AND (age_m + wait_d / 30.4375) >= 36.0, final_vendor_id, NULL) IGNORE NULLS) AS cliff_vendors
    FROM ep
""").to_dataframe()

n_breached = int(cliff_df["breached_45d"].iloc[0])
n_cliff = int(cliff_df["age_out_cliff_cases"].iloc[0])
cliff_vendors = set(cliff_df["cliff_vendors"].iloc[0]) if len(cliff_df) else set()

# Identify Eval Hubs involved in Re-Referral Bounces (the smoking-gun hubs)
bounce_hubs = set(
    edges_df.loc[edges_df["is_bounce"], "source_id"].tolist()
    + edges_df.loc[edges_df["is_bounce"], "target_id"].tolist()
)

# 2. Assign (x, y) coordinates grouped into 4 columns by node_role
role_order = [
    "Screening_Pediatric_Clinic",
    "Part_C_Intake_Coordinator",
    "Multidisciplinary_Eval_Hub",
    "EI_Therapy_Vendor",
]
role_colors = {
    "Screening_Pediatric_Clinic": "#2b6cb0",
    "Part_C_Intake_Coordinator": "#d97706",
    "Multidisciplinary_Eval_Hub": "#6b46c1",
    "EI_Therapy_Vendor": "#2f855a",
}

pos = {}
for col_idx, role in enumerate(role_order):
    sub = nodes_df[nodes_df["node_role"] == role].reset_index(drop=True)
    ys = np.linspace(0.88, 0.10, len(sub))
    for row_idx, row in sub.iterrows():
        x_offset = 0.07 if (role == "Multidisciplinary_Eval_Hub" and row_idx % 2 == 1) else 0.0
        pos[row["provider_id"]] = (col_idx + x_offset, ys[row_idx])

fig, ax = plt.subplots(figsize=(16.5, 10.5), dpi=120)

# 3. Draw directed edges (highlighting Coordinator->Eval Hub fax triggers & Stage 3 bounces)
for _, edge in edges_df.iterrows():
    x1, y1 = pos[edge["source_id"]]
    x2, y2 = pos[edge["target_id"]]
    is_coord_to_eval_fax = (
        edge["source_role"] == "Part_C_Intake_Coordinator"
        and edge["target_role"] == "Multidisciplinary_Eval_Hub"
        and edge["has_fax"]
    )

    if edge["is_bounce"]:
        # Smoking-gun Stage 3 Eval->Eval Re-Referral Bounce
        color, style, alpha, lw, rad, z = "#dc2626", "solid", 0.95, 3.0, 0.28, 4
    elif is_coord_to_eval_fax:
        # Point 1: Thick red-orange dashed edge showing which Coordinator faxes into overloaded Eval Hubs
        color, style, alpha, lw, rad, z = "#ea580c", (0, (5, 3)), 0.88, 2.5, 0.06, 3
    elif edge["has_fax"]:
        # Other paper fax handoffs
        color, style, alpha, lw, rad, z = "#f59e0b", "dashed", 0.45, 1.3, 0.03, 2
    else:
        # Clean EHR / portal handoffs
        color, style, alpha, lw, rad, z = "#0d9488", "solid", 0.35, 1.2, 0.02, 1

    ax.annotate(
        "",
        xy=(x2, y2),
        xytext=(x1, y1),
        zorder=z,
        arrowprops=dict(
            arrowstyle="->,head_width=0.38,head_length=0.48",
            color=color,
            linestyle=style,
            lw=lw,
            alpha=alpha,
            connectionstyle=f"arc3,rad={rad}",
            shrinkA=13,
            shrinkB=13,
        ),
    )

# 4. Draw CareNode vertices with explicit [Bi] vs [No Bi] & smoking-gun highlights
for _, node in nodes_df.iterrows():
    pid = node["provider_id"]
    x, y = pos[pid]
    role = node["node_role"]
    c = role_colors[role]
    wl = int(node["active_waitlist_count"])
    size = 240 + wl * 4.5

    # Highlight smoking-gun bounce Eval Hubs with a red outer ring
    is_smoking_gun = (role == "Multidisciplinary_Eval_Hub") and (pid in bounce_hubs or wl >= 63)
    edge_c = "#dc2626" if is_smoking_gun else "white"
    edge_w = 2.8 if is_smoking_gun else 1.8
    ax.scatter(x, y, s=size, color=c, edgecolors=edge_c, linewidth=edge_w, zorder=5)

    short_id = pid.replace(f"SYN-PROV-{STATE}-", f"{STATE}-")
    # Point 2: Explicitly label both [Bi] and [No Bi]
    bi_badge = "[Bi]" if node["bilingual_services_listed"] else "[No Bi]"
    cliff_tag = " ⚠36m+" if pid in cliff_vendors else ""
    label_bg = "#fef2f2" if is_smoking_gun else ("#fffbeb" if not node["bilingual_services_listed"] else "white")
    label_ec = "#f87171" if is_smoking_gun else "none"

    ax.text(
        x,
        y + 0.026,
        f"{short_id} {bi_badge} (wl={wl}){cliff_tag}",
        fontsize=8.3,
        fontweight="bold",
        color="#991b1b" if is_smoking_gun else "#1e293b",
        ha="center",
        va="bottom",
        zorder=6,
        bbox=dict(boxstyle="round,pad=0.18", fc=label_bg, ec=label_ec, lw=1.0, alpha=0.88),
    )

# 5. Stage headers, callout banner, and punchy title (Points 3 & 4)
stage_titles = [
    "Stage 1: Screening Clinics\n(Upstream [No Bi] Trigger)",
    "Stage 2: Part C Coordinators\n(Fax Routing Bottleneck)",
    "Stage 3: Multidisciplinary Eval Hubs\n(Waitlist Saturation & Bounces)",
    "Stage 4: EI Therapy Vendors\n(⚠ = 36-Month Age-Out Risk)",
]
for idx, title in enumerate(stage_titles):
    ax.text(
        idx,
        0.99,
        title,
        fontsize=11,
        fontweight="bold",
        ha="center",
        va="bottom",
        color=role_colors[role_order[idx]],
    )

legend_items = [
    Line2D([0], [0], color="#0d9488", lw=2, linestyle="solid", label="EHR Portal Handoff (Fast)"),
    Line2D([0], [0], color="#ea580c", lw=2.6, linestyle="dashed", label="Coordinator → Eval Hub Paper Fax (Root-Cause Trigger)"),
    Line2D([0], [0], color="#dc2626", lw=3.0, linestyle="solid", label="Re-Referral Bounce (Eval Hub → Eval Hub >45d Breach)"),
    mpatches.Patch(facecolor="#fef2f2", edgecolor="#dc2626", label=f"Overloaded / Bounce Eval Hub | ⚠36m+ = Age-Out Cliff ({n_cliff} toddlers age out!)"),
]
ax.legend(handles=legend_items, loc="lower center", ncol=2, frameon=True, fontsize=9.8)

# Point 4: Punchy competition title + metric subtitle
ax.set_title(
    f"{STATE_NAME} Referral Network — Re-Referral Bounce Chains (45-Day Breach)\n"
    f"{n_breached}/44 Episodes Breach Federal 45-Day Limit · {n_cliff} Toddlers Cross the 36-Month Part C Age-Out Cliff Before Therapy",
    fontsize=13.5,
    fontweight="bold",
    pad=28,
)
ax.set_xlim(-0.45, 3.45)
ax.set_ylim(-0.04, 1.05)
ax.axis("off")
plt.tight_layout()
plt.show()
````
